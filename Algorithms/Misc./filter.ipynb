🔍 filter() in Python

📌 1. Definition
	•	filter() is used to filter elements from an iterable (like list, tuple) for which a function returns True.
	•	It returns a filter object which can be converted to list/tuple.

⸻

📌 2. Syntax

filter(function, iterable)
-	function: A function that returns True or False.
-	iterable: The data to filter (e.g., list, tuple, set).

⸻

📌 3. Basic Example: Filter even numbers

✅ Using filter():

```python
def is_even(n):
    return n % 2 == 0

nums = [1, 2, 3, 4, 5, 6]
even_nums = list(filter(is_even, nums))
print(even_nums)  # Output: [2, 4, 6]
```

📌 4. Using lambda with filter()

```python
nums = [10, 15, 20, 25, 30]
result = list(filter(lambda x: x > 20, nums))
print(result)  # Output: [25, 30]
```

📌 5. Filter Strings: Keep only vowels

```python
chars = ['a', 'b', 'c', 'e', 'i', 'o', 'u']

vowels = list(filter(lambda ch: ch in 'aeiou', chars))
print(vowels)  # Output: ['a', 'e', 'i', 'o', 'u']
```

📌 6. Filter Based on Condition

```python
words = ["apple", "", "banana", "", "cherry"]
non_empty = list(filter(lambda x: x != "", words))
print(non_empty)  # Output: ['apple', 'banana', 'cherry']
```

📌 7. Comparison with for loop

🔹 Using for loop:


```python
nums = [1, 2, 3, 4]
even = []
for n in nums:
    if n % 2 == 0:
        even.append(n)
print(even)
```

🔹 Using filter():
Both return: [2, 4]

```python
even = list(filter(lambda x: x % 2 == 0, nums))
print(even)
```

📌 8. Use Case: Remove None or False values

Here, None as the function removes all falsy values.

```python
data = [None, False, 0, "", 42, "hello"]
filtered = list(filter(None, data))
print(filtered)  # Output: [42, 'hello']
```
