## ✅ Syntax

```python
{expression for item in iterable if condition}

	•	expression: The value to include in the set.
	•	item: The variable representing each element in the iterable.
	•	iterable: The sequence to loop over.
	•	condition (optional): A filter condition to include only specific elements.

🔹 Example 1: Square of numbers from 1 to 10

````python
squares = {x**2 for x in range(1, 11)}
print(squares)
````

🔹 Example 2: Unique vowels in a string

````python
sentence = "data science and algorithms"
vowels = {char for char in sentence if char in 'aeiou'}
print(vowels)
````

🔹 Example 3: Filtering even numbers

````python
even_numbers = {num for num in range(20) if num % 2 == 0}
print(even_numbers)
````


🔹 Example 4: Remove duplicates from a list

````python
numbers = [1, 2, 2, 3, 4, 4, 5]
unique_numbers = {x for x in numbers}
print(unique_numbers)
````

⚠️ Notes
-	Set comprehensions automatically eliminate duplicates.
-	The result is unordered.
-	Only immutable objects (like numbers, strings, tuples) can be stored in a set.
