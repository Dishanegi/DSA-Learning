# Reading Input from Keyboard in Python

Keyboard input is essential for creating interactive Python programs. Here's a comprehensive guide to reading user input in various formats.

## Basic Input

The `input()` function is Python's primary way to read keyboard input:

```python
# Basic input function - returns a string
name = input("Enter your name: ")
print(f"Hello, {name}!")
```

## Type Conversion

Since `input()` always returns a string, you'll often need to convert the input to other types:

```python
# Converting input to integers and floats
age = int(input("Enter your age: "))
print(f"Next year, you'll be {age + 1} years old.")

height = float(input("Enter your height in meters: "))
print(f"Your height is {height * 100} centimeters.")
```

## Multiple Inputs

You can read multiple values in one go:

```python

# Reading multiple values in one line
x, y = map(int, input("Enter two numbers separated by space: ").split())
print(f"Sum: {x + y}, Product: {x * y}")

# Reading a list of numbers
numbers = list(map(int, input("Enter several numbers separated by spaces: ").split()))
print(f"You entered: {numbers}")
print(f"Sum of numbers: {sum(numbers)}")
```

## Reading Multiple Lines

For multi-line input:

```python
# Reading multiple lines until a condition
print("Enter lines of text (type 'end' to finish):")
lines = []
while True:
    line = input()
    if line == 'end':
        break
    lines.append(line)
print(f"You entered {len(lines)} lines.")
print("Your text:")
for line in lines:
    print(f"- {line}")
```

## Error Handling with Input

Always validate user input:

```python
# Safe integer input with validation
def get_integer_input(prompt):
    while True:
        try:
            return int(input(prompt))
        except ValueError:
            print("Please enter a valid integer!")

age = get_integer_input("Enter your age: ")
print(f"You are {age} years old.")
```

## Menu-Based Input

Create interactive menus:

```python
# Creating a simple menu system
def display_menu():
    print("\n==== Main Menu ====")
    print("1. Add new item")
    print("2. View all items")
    print("3. Search items")
    print("4. Exit")
    return get_integer_input("Enter your choice (1-4): ")

# Sample usage
while True:
    choice = display_menu()
    if choice == 1:
        print("Adding new item...")
    elif choice == 2:
        print("Viewing all items...")
    elif choice == 3:
        print("Searching items...")
    elif choice == 4:
        print("Goodbye!")
        break
    else:
        print("Invalid choice. Please try again.")
```
