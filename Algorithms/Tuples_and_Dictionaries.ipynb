🧩 Tuples

✅ Simultaneous Assignment

You can assign multiple values at once using tuple unpacking:

```python
(age, name, primes) = (23, "Disha", [2, 5, 6])

print(age)
print(name)
print(primes)
```

✅ Assigning Tuples to Variables

Tuples can be used to store grouped values:

```python
point = (3.5, 4.8)
date = (16, 7, 2013)

print(point)
print(date)
```

✅ Accessing Elements and Slicing

Tuples behave like lists in terms of indexing:

```python
x_coordinate = point[0]          # 3.5
month_and_year = date[1:]       # (7, 2013)

print(x_coordinate)
print(month_and_year)
```

🚫 Immutability of Tuples

Tuples are immutable—their elements cannot be changed:

```python
# This will raise an error
date[1] = 8  
```

📦 Generalizing Lists

A list can be seen as a mapping from indices to values:

-	Mathematically:
l: \{0, 1, \dots, 5\} \rightarrow \text{integers}
l(0) = 13, l(4) = 9
-	Indices (0, 1, 2...) are keys, and l[i] are values


```python
l = [13, 24, 56, 78, 9, 10]
```

🗃️ Dictionaries

✅ Basics

Unlike lists, dictionaries allow arbitrary immutable keys:

```python
test1 = {}
test1["Dhawan"] = 24
test1["Kohli"] = 60

print(test1)
```

✅ Immutable Keys

- Valid dictionary keys: int, float, bool, str, tuple
- Invalid keys: list, dict (because they are mutable)

✅ Nesting Dictionaries

Dictionaries can contain other dictionaries:

```python
score = {
    "Test1": {"Dhawan": 84, "Kohli": 200},
    "Test2": {"Dhawan": 50}
}

# Or assign step-by-step
score = {}
score["Test1"] = {}
score["Test2"] = {}

score["Test1"]["Dhawan"] = 36
score["Test1"]["Kohli"] = 36
score["Test2"]["Dhawan"] = 40
```

```python
# Example

score = {}
score["Test1"] = {}
score["Test2"] = {}

score["Test1"]["Dhawan"] = 36
score["Test2"]["Dhawan"] = 40
score["Test1"]["Kohli"] = 36

print(score)
```

✅ Accessing and Iterating

```python
# Print entire dictionary
print(score)

# Iterate through keys
for k in score.keys():
    print(k, score[k])

# Sorted iteration
for k in sorted(score.keys()):
    print(k, score[k])

# Convert to list explicitly if needed
key_list = list(score.keys())

print(key_list)
```

✅ Working with Values

```python
total = 0
for val in test1.values():
    total += val

print(total)
```

✅ Checking Key Membership

```python
total = {}
for player in ["Dhawan", "Kohli"]:
    total[player] = 0
    for match in score:
        if player in score[match]:
            total[player] += score[match][player]

print(total)
```
