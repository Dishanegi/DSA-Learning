✨ Quicksort: A Fast & Elegant Sorting Algorithm

Quicksort is one of the most efficient and widely used sorting algorithms, leveraging the power of divide and conquer. It operates by selecting a pivot element and rearranging the array such that all values less than or equal to the pivot appear before it, and all greater values appear after. This process, known as partitioning, is then recursively applied to the sub-arrays on either side of the pivot.

Its average-case time complexity of O(n log n) makes it a go-to choice for large datasets, and its in-place variant offers excellent space efficiency.

📘 Functional Implementation (Returns a New Sorted List)

```python
def quickSort(arr):
    # Base case: an array of 0 or 1 element is already sorted
    if len(arr) <= 1:
        return arr
    
    # Select pivot (here, first element)
    pivot = arr[0]
    
    # Partition: elements <= pivot to the left, > pivot to the right
    left = [x for x in arr[1:] if x <= pivot]
    right = [x for x in arr[1:] if x > pivot]
    
    # Recursively sort and combine
    return quickSort(left) + [pivot] + quickSort(right)

# Example usage
print("Sorted array (functional):", quickSort([2, 4, -1, 0, 8, 10, 1, 0, 0]))
```

🔧 In-Place Quicksort Implementation

```python
def quick_sort(A, l, r):
    # Base case: sub-array of size 1 or 0
    if r - l <= 1:
        return

    # Yellow pointer marks the split point
    yellow = l + 1

    # Green scans through the sub-array
    for green in range(l + 1, r):
        if A[green] <= A[l]:
            A[yellow], A[green] = A[green], A[yellow]
            yellow += 1
    
    # Place pivot (A[l]) in its correct position
    A[l], A[yellow - 1] = A[yellow - 1], A[l]

    # Recursively apply to left and right sub-arrays
    quick_sort(A, l, yellow - 1)
    quick_sort(A, yellow, r)

# Example usage
arr = [2, 4, -1, 0, 8, 10, 1, 0, 0]
quick_sort(arr, 0, len(arr))
print("Sorted array (in-place):", arr)
```

## 🔍 Quicksort Algorithm – Analysis and Insights

Quicksort is a highly efficient **divide-and-conquer** sorting algorithm that is widely used due to its average-case speed and in-place sorting ability.

---

### 🧠 How Quicksort Works

1. **Choose a Pivot** – typically the first, last, or a random element.
2. **Partition** the array into:
   - Elements **less than or equal to** the pivot.
   - Elements **greater than** the pivot.
3. **Recursively** apply the same strategy to the left and right subarrays.

---

### ⏱️ Time Complexity

| Case       | Scenario                                                        | Time Complexity |
|------------|------------------------------------------------------------------|-----------------|
| **Best**   | Pivot splits array evenly every time                             | `O(n log n)`    |
| **Average**| Pivot splits are reasonably balanced on average                  | `O(n log n)`    |
| **Worst**  | Pivot is always the smallest/largest (e.g., sorted input)        | `O(n²)`         |

> 🔸 In practice, **randomized pivoting** is used to avoid the worst case.

---

### 💾 Space Complexity

| Case       | Stack Depth             | Space Complexity |
|------------|--------------------------|------------------|
| **Best**   | `log n` recursive calls  | `O(log n)`       |
| **Worst**  | `n` recursive calls      | `O(n)`           |

- **Quicksort is in-place** (i.e., it doesn’t use extra arrays).
- Space is consumed by the **recursive call stack**.

---

### ⚖️ Stability

- ❌ **Not stable** by default (equal elements may not retain order).
- ✅ Can be made stable with extra space, but that removes the in-place benefit.

---

### 📊 Performance Summary

| Property         | Value          |
|------------------|----------------|
| **Best Time**    | `O(n log n)`   |
| **Average Time** | `O(n log n)`   |
| **Worst Time**   | `O(n²)`        |
| **Space**        | `O(log n)`     |
| **In-place**     | ✅ Yes          |
| **Stable**       | ❌ No           |

---

### 🧪 Example: Sorting `[5, 2, 8, 3, 1]`

1. Pivot = 5  
2. Partition: `[2, 3, 1]` and `[8]`
3. Recurse on `[2, 3, 1]` → sorted as `[1, 2, 3]`
4. Final: `[1, 2, 3, 5, 8]`

---

### 🏁 Final Thoughts

Quicksort strikes a balance between **speed**, **space efficiency**, and **simplicity**. While it's not always the best in the worst-case scenario, **with random pivots or hybrid implementations**, it is one of the fastest and most practical algorithms in real-world systems (like Python’s `Timsort`, which uses quicksort + mergesort techniques).

> 📌 Use Quicksort when you need speed and can tolerate occasional instability.
