## 🔸 `del` Keyword
`del` is used to delete **variables**, **list elements**, or **object attributes** from memory.

### Delete a variable
```python
x = 10
del x
# print(x)  # This would raise NameError
```

### Delete from a list
```python
lst = [1, 2, 3, 4]
del lst[2]  # Removes the element at index 2
print(lst)  # Output: [1, 2, 4]
```

### Delete object attributes
```python
class Person:
    def __init__(self):
        self.name = "Alice"

p = Person()
del p.name
# print(p.name)  # This will raise AttributeError
```

## 🔸 `None` Keyword
`None` is a special constant in Python that represents the **absence of a value** or a **null value**.

### Basic Usage
```python
x = None
if x is None:
    print("x has no value")
```

### Function returning None
```python
def greet():
    print("Hello")

result = greet()  # prints Hello
print(result)     # prints None
```

`None` is commonly used to initialize variables that will later hold other types of values.
