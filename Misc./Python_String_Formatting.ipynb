# 🧵 Python String Formatting

Python supports multiple ways to format strings. Starting from Python 3.6, **f-strings** became the preferred approach.

## 🔹 F-Strings (Python 3.6+)
To create an f-string, prefix the string with `f`:
```python
txt = f"The price is 49 dollars"
print(txt)
```

````python
price = 59
txt = f"The price is {price} dollars"
print(txt)

# Display with 2 decimals
txt = f"The price is {price:.2f} dollars"
print(txt)

# Inline values
txt = f"The price is {95:.2f} dollars"
print(txt)
````

### Operations in F-Strings


````python
txt = f"The price is {20 * 59} dollars"
print(txt)

price = 59
tax = 0.25
txt = f"The price is {price + (price * tax)} dollars"
print(txt)

price = 49
txt = f"It is very {'Expensive' if price > 50 else 'Cheap'}"
print(txt)
````

### Functions in F-Strings


````python
fruit = "apples"
txt = f"I love {fruit.upper()}"
print(txt)

def myconverter(x):
    return x * 0.3048

txt = f"The plane is flying at a {myconverter(30000)} meter altitude"
print(txt)
````

### More Modifiers

**Formatting Types Summary**:
- `:<` : Left-align
- `:>` : Right-align
- `:^` : Center-align
- `:=` : Sign to left
- `:+` : Plus for positive/negative
- `:-` : Minus only
- `:,` : Comma separator
- `:_` : Underscore separator
- `:b/o/x/X` : Binary/Octal/Hex
- `:e/E/f/F/g/G/n/%` : Scientific, fixed-point, percentage etc.

````python
price = 59000
txt = f"The price is {price:,} dollars"
print(txt)
````

## 🔹 Using `format()` Method
Before f-strings, Python used `.format()`:


````python
price = 49
txt = "The price is {} dollars"
print(txt.format(price))

txt = "The price is {:.2f} dollars"
print(txt.format(price))
````

### Multiple Values



````python
quantity = 3
itemno = 567
price = 49
myorder = "I want {} pieces of item number {} for {:.2f} dollars."
print(myorder.format(quantity, itemno, price))

# Using index numbers
myorder = "I want {0} pieces of item number {1} for {2:.2f} dollars."
print(myorder.format(quantity, itemno, price))

# Repeating values
age = 36
name = "John"
txt = "His name is {1}. {1} is {0} years old."
print(txt.format(age, name))
````

### Named Indexes

````python
myorder = "I have a {carname}, it is a {model}."
print(myorder.format(carname = "Ford", model = "Mustang"))
````
