# 🧵 String Manipulation Methods in Python (With Explanations)

## 1. `strip()`, `lstrip()`, `rstrip()`
- **`strip()`** removes whitespace (or specified characters) from **both ends** of a string.
- **`lstrip()`** removes from the **left**.
- **`rstrip()`** removes from the **right**.

```python
s = "   Hello World!  "
print(s.strip())    # "Hello World!"
print(s.lstrip())   # "Hello World!  "
print(s.rstrip())   # "  Hello World!"
```

## 2. `min()` and `max()`
- **`min()`** returns the character with the **lowest ASCII value**.
- **`max()`** returns the character with the **highest ASCII value**.

```python
s = "python"
print(min(s))  # "h"
print(max(s))  # "y"
```

## 3. `replace()`
- **`replace(old, new)`** replaces **all occurrences** of `old` with `new`.

```python
s = "Hello World"
print(s.replace("World", "Python"))  
```

## 4. `expandtabs(tabsize)`
- Replaces **tab characters (\t)** with spaces.
- `tabsize` defines the number of spaces per tab.

```python
s = "Python\tis\tawesome"
print(s.expandtabs(4))  
```

## 5. `find()` and `index()`
- **`find()`** returns the **first index** of a substring. Returns **-1 if not found**.
- **`index()`** is similar but **raises an error** if not found.

```python
s = "Data Science"
print(s.find("a"))    # 1
print(s.index("a"))   # 1
```

## 6. `split()` and `join()`
- **`split(sep)`** splits the string at the separator and returns a list.
- **`join(iterable)`** joins elements using the string as a separator.

```python
s = "one,two,three"
parts = s.split(",")
print(parts)               
print(",".join(parts))     
```

## 7. `upper()`, `lower()`, `title()`, `swapcase()`, `capitalize()`
- **`upper()`**: Converts to uppercase.
- **`lower()`**: Converts to lowercase.
- **`title()`**: Capitalizes first letter of each word.
- **`swapcase()`**: Swaps case of each character.
- **`capitalize()`**: Capitalizes only the first character.

```python
s = "hello world"
print(s.upper())        # "HELLO WORLD"
print(s.lower())        # "hello world"
print(s.title())        # "Hello World"
print(s.swapcase())     # "HELLO WORLD"
print(s.capitalize())   # "Hello world"
```

## 8. `center()`, `ljust()`, `rjust()`
- Aligns the string with **padding** characters.
- **`center(width, fillchar)`** centers it.
- **`ljust()`** aligns left.
- **`rjust()`** aligns right.

```python
s = "Python"
print(s.center(10, "*"))  # "**Python**"
print(s.ljust(10, "-"))   # "Python----"
print(s.rjust(20, "."))   # "....Python"
```

## 9. `isalpha()` and `isnumeric()`
- **`isalpha()`** returns `True` if all characters are letters.
- **`isnumeric()`** returns `True` if all characters are numbers.

```python
print("abc".isalpha())     # True
print("123".isnumeric())   # True
print("abc123".isalpha())  # False
print("12.3".isnumeric())  # False
```
