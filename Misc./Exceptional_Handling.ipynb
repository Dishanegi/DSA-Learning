
# Python Exception Handling

Exception handling in Python allows you to manage errors gracefully during program execution.

## Core Exception Handling Components
- `try`: Block of code that might raise exceptions
- `except`: Handles specific exceptions
- `else`: Executes if no exceptions occur
- `finally`: Always executes, regardless of exceptions

## Common Exception Types
- `ZeroDivisionError`: Division by zero
- `FileNotFoundError`: Cannot find specified file
- `ValueError`: Incorrect value type
- `TypeError`: Incorrect data type
- `IndexError`: Index out of range
- `KeyError`: Key not found in dictionary

```python
# Basic try-except structure
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero!")
    
# More comprehensive example
try:
    file = open("nonexistent_file.txt", "r")
except FileNotFoundError:
    print("The file doesn't exist!")
except Exception as e:
    print(f"An error occurred: {e}")
else:
    print("File opened successfully!")
    file.close()
finally:
    print("Exception handling complete")

# Raising exceptions
def check_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative")
    if not isinstance(age, int):
        raise TypeError("Age must be an integer")
    return f"Age validated: {age}"

# Custom exception example
class InsufficientFundsError(Exception):
    """Raised when a withdrawal would result in a negative balance"""
    def __init__(self, balance, withdrawal_amount):
        self.balance = balance
        self.withdrawal_amount = withdrawal_amount
        self.message = f"Cannot withdraw ${withdrawal_amount}. Only ${balance} available."
        super().__init__(self.message)

def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientFundsError(balance, amount)
    return balance - amount
```
