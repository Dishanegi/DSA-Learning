🔹 Functions in Python

A function in Python is a reusable block of code that performs a specific task. It helps to break your program into smaller, manageable parts and allows for code reuse.

⸻

✅ Why Use Functions?
	•	Avoid code repetition
	•	Improve readability and maintainability
	•	Enable modular programming

🔹 Defining a Function

You define a function using the def keyword:

This function takes one argument name and prints a greeting.

```python
def greet(name):
    print(f"Hello, {name}!")
```

🔹 Calling a Function

```python
greet("Alice")
# Output: Hello, Alice!
```

🔹 Return Values

Use return to send back a result:

```python
def add(a, b):
    return a + b

result = add(3, 5)
print(result)  # Output: 8
```

🔹 Function Arguments

Python supports various types of arguments:
1.	Positional Arguments


```python
def power(base, exponent):
    return base ** exponent

print(power(2, 3))  # Output: 8
```

2.	Keyword Arguments

```python
print(power(exponent=3, base=2))  # Output: 8
```

3.	Default Arguments
- Default alue is provided in function definition
- If parameter is moitted, default value is used
- Order of the arguments are important
    - Default avlue must be available ad definition time
    - def Quicksort(a,l=0,len(A)): does not work

```python
def greet(name="Guest"):
    print(f"Hello, {name}!")

greet()             # Output: Hello, Guest!
greet("Charlie")    # Output: Hello, Charlie!
```

4.	Variable-length Arguments
    1.	args (tuple of positional arguments):
    2.	kwargs (dictionary of keyword arguments):




```python
def total(*args):
    return sum(args)

print(total(1, 2, 3))  # Output: 6
```

```python
def display(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

display(name="Alice", age=30)
```

🔹 Lambda Functions (Anonymous Functions)

Used for short, one-line functions:

```python
square = lambda x: x * x
print(square(4))  # Output: 16
```

🔹 Scope of Variables
1.	Local scope: inside the function
2.	Global scope: outside the function

```python
x = 5  # Global

def show():
    x = 10  # Local
    print(x)

show()       # Output: 10
print(x)     # Output: 5
```
