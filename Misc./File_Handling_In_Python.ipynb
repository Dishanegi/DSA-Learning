# 📁 File Handling in Python

File handling refers to the process of performing operations on a file such as creating, opening, reading, writing and closing it, through a programming interface. It involves managing the data flow between the program and the file system on the storage device, ensuring that data is handled safely and efficiently.

## 🔓 Opening a File in Python
To open a file we can use `open()` function, which requires file path and mode as arguments:
```python
# Open the file and read its contents
with open('geeks.txt', 'r') as file:
    content = file.read()
    print(content)
```
This code opens file named `geeks.txt`.

## 📃 File Modes in Python
| Mode | Description | Behavior |
|------|-------------|----------|
| `r` | Read-only mode. | Opens the file for reading. File must exist. |
| `rb` | Read-only in binary mode. | Reads binary data. File must exist. |
| `r+` | Read and write mode. | File must exist. |
| `rb+` | Read and write binary. | File must exist. |
| `w` | Write mode. | Overwrites or creates new file. |
| `wb` | Write in binary mode. | Overwrites or creates binary file. |
| `w+` | Write and read. | Truncates or creates new file. |
| `wb+` | Write/read binary. | Truncates or creates binary file. |
| `a` | Append mode. | Appends to file or creates new. |
| `ab` | Append binary. | Appends binary or creates. |
| `a+` | Append and read. | Appends and reads or creates. |
| `ab+` | Append/read binary. | Appends/reads binary or creates. |
| `x` | Exclusive creation. | Fails if file exists. |
| `xb` | Exclusive binary. | Fails if file exists. |
| `x+` | Exclusive read/write. | Fails if file exists. |
| `xb+` | Exclusive binary read/write. | Fails if file exists. |

## 📖 Reading a File
```python
file = open("geeks.txt", "r")
content = file.read()
print(content)
file.close()
```
**Output:**
```
Hello world
GeeksforGeeks
123 456
```

````python
# Reading a file in binary mode
file = open("geeks.txt", "rb")
content = file.read()
print(content)
file.close()
````

## ✍ Writing to a File
```python
file = open("geeks.txt", "w")
file.write("Hello, World!")
file.close()
```

````python
# Appending to a file
file = open('geek.txt', 'a')
file.write("This will add this line")
file.close()
````

## 🔒 Closing a File
```python
file = open("geeks.txt", "r")
# Perform file operations
file.close()
```

````python
# Using 'with' statement for automatic handling
with open("geeks.txt", "r") as file:
    content = file.read()
    print(content)
````

## ⚠ Handling Exceptions While Closing
```python
try:
    file = open("geeks.txt", "r")
    content = file.read()
    print(content)
finally:
    file.close()
```

## ✅ Advantages of File Handling in Python
- **Versatility**: Perform a wide range of operations.
- **Flexibility**: Supports many file types.
- **User-friendly**: Easy to read and manipulate.
- **Cross-platform**: Works on Windows, Linux, Mac.

## ⚠️ Disadvantages of File Handling in Python
- **Error-prone**: If not carefully written.
- **Security risks**: Especially with untrusted user input.
- **Complexity**: Advanced formats can be tricky.
- **Performance**: May be slower for large/complex data.
