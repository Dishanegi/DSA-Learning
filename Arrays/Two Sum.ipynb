# Two Sum

## 📝 Problem Statement

Given an array of integers `nums` and an integer `target`, return indices of the two numbers such that they add up to `target`.

You may assume that each input would have **exactly one solution**, and you may not use the same element twice.

You can return the answer in any order.

---

## 📊 Examples

### Example 1:
```
Input: nums = [2,7,11,15], target = 9
Output: [0,1]
```
**Explanation:** Because nums[0] + nums[1] == 9, we return [0, 1].

### Example 2:
```
Input: nums = [3,2,4], target = 6
Output: [1,2]
```

### Example 3:
```
Input: nums = [3,3], target = 6
Output: [0,1]
```

---

## ⚠️ Constraints

- 2 ≤ nums.length ≤ 10⁴
- -10⁹ ≤ nums[i] ≤ 10⁹
- -10⁹ ≤ target ≤ 10⁹
- Only one valid answer exists.

#Brute Force Method of using 2 For loops

````python
def twoSumBF(nums,target):
    for i in range(len(nums)):
        for j in range(i+1,len(nums)):
            diff = target - nums[i]
            if diff == nums[j]:
                return [i,j]
    return []

twoSumBF([15,11,2,7],9)
````

````python
# 2 Pass Hash Table
def twoSumPassTable(nums,target):
    hashmap = {}
    for i in range(len(nums)):
        hashmap[nums[i]] = i
    for i in range(len(nums)):
        diff = target - nums[i]
        if diff in hashmap and hashmap[diff] !=i:
            return [i,hashmap[diff]]
    return []

twoSumPassTable([15,11,2,7],9)
````

````python
# One-pass Hash Table
def twoSumOnePassTable(nums,target):
    hashmap={}
    for i in range(len(nums)):
        diff = target - nums[i]
        if diff in hashmap:
            return (i,hashmap[diff])
        hashmap[nums[i]] = i
    return []

twoSumOnePassTable([15,11,2,7],9)
````

````python
def best_poker_hand(cards):
    """
    Determine the best poker hand from 5 cards.
    
    Args:
        cards (list): A list of 5 card strings, each with a number and a symbol
                     (e.g., ['10a', '4a', '2a', '7a', '9a'])
    
    Returns:
        str: The name of the best poker hand ("Flush", "Three of a Kind", or "Pair")
             or None if no valid hand is found
    """
    # Extract symbols and numbers from each card
    symbols = [card[-1] for card in cards]  # Get the last character of each card as the symbol
    numbers = [card[:-1] for card in cards]  # Get all characters except the last as the number
    
    # Check for Flush (all symbols are the same)
    if len(set(symbols)) == 1:
        return "Flush"
    
    # Count occurrences of each number
    number_counts = {}
    for num in numbers:
        if num in number_counts:
            number_counts[num] += 1
        else:
            number_counts[num] = 1
    
    # Check for Three of a Kind (3 cards with the same number)
    for num, count in number_counts.items():
        if count >= 3:
            return "Three of a Kind"
    
    # Check for Pair (2 cards with the same number)
    for num, count in number_counts.items():
        if count >= 2:
            return "Pair"
    
    # No valid hand found
    return None

best_poker_hand(['10a', '4a', '2a', '7a', '9a'])
````

# 🌟 Lessons Learned

### ⏱️ Performance Analysis
| Metric | Complexity | Description |
|--------|------------|-------------|
| **Time Complexity** | `O(n)` | We traverse the array only once, achieving linear time performance |
| **Space Complexity** | `O(n)` | In the worst case, our hash map stores n elements |

## 💡 Key Insights

### 🔍 Hash Maps for Lookup
```
O(n²) → O(n)
```
Using a hash map allows us to check if an element exists in `O(1)` time, dramatically improving efficiency over the `O(n)` linear search.

### 🧩 Complement Calculation
```python
diff = target - n
if diff in prevdict:  # O(1) lookup!
    return [prevdict[diff], i]
```
Instead of checking every pair, we calculate exactly what value we need (the complement) and check if we've seen it before.

### ⚖️ Space-Time Trade-off
```
Space ↑, Time ↓
```
We trade space (using a hash map) for time (reducing from `O(n²)` to `O(n)`), a classic algorithmic optimization technique.

### 🐍 Pythonic Elegance
```python
for i, n in enumerate(nums):  # Clean and readable
```
Using `enumerate()` gives us cleaner, more readable code while maintaining optimal performance.
